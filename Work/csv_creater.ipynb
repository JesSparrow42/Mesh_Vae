```python
%%writefile pt_export.py
import os, re
from playwright.sync_api import sync_playwright

EMAIL = os.environ["PT_EMAIL"]
PASSWORD = os.environ["PT_PASSWORD"]

with sync_playwright() as p:
    browser = p.chromium.launch(headless=True)  # set False to watch
    context = browser.new_context(accept_downloads=True)
    page = context.new_page()

    page.goto("https://new.tcgpowertools.com/", wait_until="domcontentloaded")

    page.get_by_label(re.compile("email", re.I)).fill(EMAIL)
    page.get_by_label(re.compile("password", re.I)).fill(PASSWORD)
    page.get_by_role("button", name=re.compile("log in|login|sign in", re.I)).click()

    page.goto("https://new.tcgpowertools.com/stock-pricing", wait_until="networkidle")

    with page.expect_download() as dl:
        page.get_by_role("button", name=re.compile("export", re.I)).click()

    dl.value.save_as("powertools-stock.csv")
    browser.close()

print("Saved: powertools-stock.csv")


```
