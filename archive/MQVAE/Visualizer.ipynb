```python
from pathlib import Path
import numpy as np
import open3d as o3d

exp_name = "REACT_vae_boson"
base = Path("checkpoints") / exp_name / "reconstructions"

def show_o3d_mesh(path: Path, label: str):
    if not path.exists():
        print(f"[{label}] MISSING: {path}")
        return

    mesh = o3d.io.read_triangle_mesh(str(path))
    mesh.compute_vertex_normals()

    v = np.asarray(mesh.vertices)
    f = np.asarray(mesh.triangles)
    print(f"[{label}] {path}")
    print(f"  verts: {v.shape}, faces: {f.shape}")

    # Opens an interactive window – close it to continue
    o3d.visualization.draw_geometries([mesh], window_name=label)
```

```python
idx = 0  # change if you have more

orig_path  = base / f"original_{idx}.obj"
simp_path  = base / f"simplified_{idx}.obj"
recon_path = base / f"recon_{idx}.obj"
ref_path   = base / f"recon_{idx}_redetail.obj"

show_o3d_mesh(orig_path,  "original")
show_o3d_mesh(simp_path,  "simplified")
show_o3d_mesh(recon_path, "recon")
show_o3d_mesh(ref_path,   "recon_redetail")
```

```python
import numpy as np
import open3d as o3d
import torch

orig_path = r"C:\Users\JonatanEmilSvendsen\OneDrive - Sparrow Quantum Aps\Desktop\MQVAE\checkpoints\REACT_vae_debug_gauss\reconstructions\SPRING2257_orig_points.ply"
recon_path = r"C:\Users\JonatanEmilSvendsen\OneDrive - Sparrow Quantum Aps\Desktop\MQVAE\checkpoints\REACT_vae_debug_gauss\reconstructions\SPRING2257_recon_points.ply"

# ---- load one orig point cloud
pcd = o3d.io.read_point_cloud(orig_path)
P = np.asarray(pcd.points, dtype=np.float32)  # (N,3)
print("Loaded orig points:", P.shape, P.dtype)

# ---- original order -> tensor [1,3,N]
x1 = torch.from_numpy(P.T).unsqueeze(0)

# ---- shuffled order -> tensor [1,3,N]
perm = np.random.permutation(P.shape[0])
x2 = torch.from_numpy(P[perm].T).unsqueeze(0)

# ---- Correct set-equality check (row-wise / point-wise)
def sort_points_rows(A: np.ndarray) -> np.ndarray:
    A = np.ascontiguousarray(A)
    idx = np.lexsort((A[:, 2], A[:, 1], A[:, 0]))  # sort by x then y then z
    return A[idx]

same_set = np.array_equal(sort_points_rows(P), sort_points_rows(P[perm]))
print("Same set, different order (CORRECT CHECK):", same_set)

# (optional) sanity: tensors differ in order
print("Same tensor (order-sensitive):", torch.allclose(x1, x2))
```

```python
pcd_orig = o3d.io.read_point_cloud(orig_path)
pcd_recon = o3d.io.read_point_cloud(recon_path)

# Open3D gives nearest-neighbor distances one-way; do both ways for a Chamfer-like score
d1 = np.asarray(pcd_orig.compute_point_cloud_distance(pcd_recon), dtype=np.float32)
d2 = np.asarray(pcd_recon.compute_point_cloud_distance(pcd_orig), dtype=np.float32)

print("Orig -> Recon mean dist:", float(d1.mean()), " 95%:", float(np.percentile(d1, 95)))
print("Recon -> Orig mean dist:", float(d2.mean()), " 95%:", float(np.percentile(d2, 95)))
print("Chamfer-like mean:", float(d1.mean() + d2.mean()))
```

```python
import numpy as np
import open3d as o3d

orig = np.asarray(o3d.io.read_point_cloud(orig_path).points, dtype=np.float32)
reco = np.asarray(o3d.io.read_point_cloud(recon_path).points, dtype=np.float32)

def spread(P, name):
    std = P.std(axis=0)
    ext = P.max(axis=0) - P.min(axis=0)
    print(name)
    print("  std   :", std)
    print("  extent:", ext)

spread(orig, "orig")
spread(reco, "recon")
```
